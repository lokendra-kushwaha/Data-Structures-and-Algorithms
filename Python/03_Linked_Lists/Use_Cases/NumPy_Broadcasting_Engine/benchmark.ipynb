```python
# ==========================================
# 2. PYTHON LIST IMPLEMENTATION
# ==========================================
def get_broadcast_shape_python_list(shape1: tuple, shape2: tuple) -> tuple | str:
    s1 = list(shape1)
    s2 = list(shape2)

    n = len(s1)
    m = len(s2)

    # O(N) Memory Shifting Padding
    if n > m:
        for _ in range(n - m):
            s2.insert(0, 1)
    elif n < m:
        for _ in range(m - n):
            s1.insert(0, 1)

    new_shape = []
    for i in range(max(n, m)):
        if s1[i] == s2[i] or s1[i] == 1 or s2[i] == 1:
            new_shape.append(max(s1[i], s2[i]))
        else:
            return '❌ Error: Cannot broadcast'
            
    return tuple(new_shape)
```

```python
import time

# ==========================================
# 3. THE BENCHMARKING ENGINE
# ==========================================
def run_benchmark():
    print("🚀 Initiating Broadcasting Validator Benchmark...\n")
    
    # TEST CASE 1: Normal Real-World Matrices (e.g., 4D and 1D)
    shape_a = (15, 3, 5, 4)
    shape_b = (4,)
    
    print("--- TEST 1: Normal Shapes ---")
    
    # Benchmark Linked List
    start_time = time.perf_counter()
    for _ in range(10000):  
        get_broadcast_shape_linked_list(shape_a, shape_b)
    ll_time = time.perf_counter() - start_time
    print(f"🔗 Linked List Engine Time:  {ll_time:.5f} seconds")

    # Benchmark Python List
    start_time = time.perf_counter()
    for _ in range(10000):
        get_broadcast_shape_python_list(shape_a, shape_b)
    pl_time = time.perf_counter() - start_time
    print(f"🐍 Python List Engine Time: {pl_time:.5f} seconds")
    print("-" * 30)

    # TEST CASE 2: The "Extreme Stress Test" 
    # Forcing massive padding to expose the O(N^2) shifting flaw in Python Lists
    extreme_shape_a = tuple([5] * 2000) # 2000 Dimensions
    extreme_shape_b = (5,)              # 1 Dimension (Requires 1999 paddings)
    
    print(f"\n--- TEST 2: Extreme Stress Test (Padding {len(extreme_shape_a) - 1} Dimensions) ---")
    
    # Benchmark Linked List
    start_time = time.perf_counter()
    get_broadcast_shape_linked_list(extreme_shape_a, extreme_shape_b)
    ll_stress_time = time.perf_counter() - start_time
    print(f"🔗 Linked List Engine Time:  {ll_stress_time:.5f} seconds")

    # Benchmark Python List
    start_time = time.perf_counter()
    get_broadcast_shape_python_list(extreme_shape_a, extreme_shape_b)
    pl_stress_time = time.perf_counter() - start_time
    print(f"🐍 Python List Engine Time: {pl_stress_time:.5f} seconds")
    print("-" * 30)

# Run the benchmark
run_benchmark()
```
